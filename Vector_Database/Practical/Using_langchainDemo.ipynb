```python
# 1. Necessary Modules Import Karein
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma

# ==========================================
# STEP 1: Content / Data Tayyar Karna
# ==========================================
# Farz karein yeh hamari ek chhoti si document ya text file hai
raw_text = """
Artificial Intelligence (AI) is a broad field of computer science focused on building smart machines.
Machine Learning (ML) is a subset of AI that enables systems to learn from data automatically.
Deep Learning uses multi-layer neural networks to analyze complex patterns like speech and images.
ChromaDB is a fast, open-source vector database used for embedding storage in AI pipelines.
LangChain is a popular framework that integrates LLMs with external data sources like vector databases.
"""

# ==========================================
# STEP 2: Text Chunking (LangChain Splitter)
# ==========================================
# LangChain ka TextSplitter lambe text ko chhote paragraphs/chunks mein todta hai
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=120,      # Har chunk mein max kitne characters honge
    chunk_overlap=20     # Context miss na ho, isliye 20 chars pichle chunk se match honge
)

# Text ko chunks (Document objects) mein convert karein
docs = text_splitter.create_documents([raw_text])

print(f"Total Chunks Ban Gaye: {len(docs)}")
print("Pehla Chunk:", docs[0].page_content)
print("-" * 50)

# ==========================================
# STEP 3: Embedding Model Set Up Karna
# ==========================================
# Free open-source embedding model (wahi model jo Chroma ne auto-download kiya tha)
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# ==========================================
# STEP 4: ChromaDB Vector Store Banayein
# ==========================================
# LangChain khud hi har chunk ka vector banayega aur ChromaDB mein save kar dega
vectorstore = Chroma.from_documents(
    documents=docs,
    embedding=embeddings,
    collection_name="langchain_simple_demo"
)

# ==========================================
# STEP 5: Smart Search (Retrieval)
# ==========================================
query = "Which tool helps connect LLMs to external databases?"

# Sab se zyada matching chunk (k=1) dhoondein
search_results = vectorstore.similarity_search(query, k=1)

print("\n--- SEARCH RESULT ---")
print("User Query:", query)
print("Best Match Found:", search_results[0].page_content)
```
